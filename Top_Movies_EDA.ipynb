- Will convert Vote_average to Ordinal words

```python
def categorize_col(df,col,labels):
    bins = df[col].quantile([0,0.25,0.5,0.75,1.0]).values

    df[col] = pd.cut(df[col],bins = bins,labels = labels,include_lowest = True)
    return df

labels = ["not_popular", "below_average", "average", "popular"]
```

```python
df = categorize_col(df,"Vote_Average",labels)
```

- What movie got the highest popularity? what's its genre?

```python
top10_movies = df.groupby("Title")["Popularity"].agg("sum").sort_values(ascending = False).head(10).reset_index()
a = top10_movies["Title"]
```

```python
genre = df[df["Title"]==a[0]]["Genre"]
genre.tolist()
```

```python
plt.figure(figsize=(14,6))
sns.barplot(data = top10_movies,x = "Title",y = "Popularity")
plt.xticks(rotation = 90)
plt.show()
```

```python
print(f"The movie that highest popularity is:\n {top10_movies.iloc[0]["Title"]} \nand its genres are: \n{genre.tolist()} \n")
```

- What movie got the lowest popularity? what's its genre?

```python
lowest10_movies = df.groupby("Title")["Popularity"].agg("sum").sort_values(ascending = True).head(10).reset_index()
lowest10_movies
```

```python
plt.figure(figsize=(14,6))
sns.barplot(data = lowest10_movies,x = "Title",y = "Popularity")
plt.grid(True)
plt.xticks(rotation = 90)
plt.show()
```

```python
b =lowest10_movies.iloc[0]["Title"]
b
genre2 = df[df["Title"]==b]["Genre"]

```

```python
print(f"The movie that has lowest popularity is:\n {lowest10_movies.iloc[0]["Title"]} \nand its genres are: \n{genre2.tolist()} \n")
```
